## Normality Test

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.normality` - Tests for normal distribution
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.stats.normality(
    dv='age',
    between='class',
    method='jarque_bera',
    show_qqplot=True
)
````

## Levene's Test

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.levene` - Tests equality of variances across groups.
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.stats.levene(
    dv='age',
    between='class'
)
````

## T-Test

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.ttest` - Compares means between two groups using independent samples t-test.
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.stats.ttest(
    dv='age',
    between='alive',
)
````

## Mann-Whitney U Test

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.mwu` - Non-parametric test for comparing two independent groups.
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.stats.mwu(
    dv='age',
    between='alive'
)
````

## ANOVA

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.anova` - One-way analysis of variance for comparing means across multiple groups.
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.stats.anova(
    dv='age',
    between='class'
)
````

## Kruskal-Wallis Test

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.kruskal` - Non-parametric alternative to one-way ANOVA.
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.stats.kruskal(
    dv='age',
    between='class'
)
````

## Chi-Square Test

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.chi2_independence` - Tests categorical variable independence
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.stats.chi2_independence(
    x='class',
    y='alive',
    test='pearson'
)
````

## Bootstrap

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.bootstrap` - Computes bootstrap confidence intervals
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.stats.bootstrap(
    dv='age',
    between='alive',
    reference_group='no',
    statistic='mean_diff',
    plot=True
)
````

## Linear Regression

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.ols` - Ordinary least squares regression
```

````python
from frameon import load_dataset, FrameOn as fo

diamonds = fo(load_dataset('diamonds'))
diamonds.stats.ols(
    formula='price ~ carat + C(cut, Treatment("Ideal"))',
    cov_type='HC3',
    show_plots=True
)
````

## Robust Linear Model

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.rlm` - Regression with robust standard errors
```

````python
from frameon import load_dataset, FrameOn as fo

diamonds = fo(load_dataset('diamonds'))
diamonds.stats.rlm(
    formula='price ~ carat + C(cut, Treatment("Ideal"))',
)
````

## Generalized Linear Model

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.glm` - Generalized linear regression
```

````python
from frameon import load_dataset, FrameOn as fo
import statsmodels.api as sm

titanic = fo(load_dataset('titanic'))
titanic.stats.glm(
    formula='survived ~ age + sex + pclass',
    family=sm.families.Binomial()
)
````

## Quantile Regression

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.quantreg` - Quantile regression
```

````python
from frameon import load_dataset, FrameOn as fo

diamonds = fo(load_dataset('diamonds'))
diamonds.stats.quantreg(
    formula='price ~ carat + C(cut, Treatment("Ideal"))',
    q=0.95
)
````

## Ordered Model

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.ordered_model` - Regression for ordered categorical dependent variables.
```

````python
from frameon import load_dataset, FrameOn as fo
import pandas as pd

diamonds = fo(load_dataset('diamonds'))
cut_order = ['Fair', 'Good', 'Very Good', 'Premium', 'Ideal']
diamonds['cut'] = pd.Categorical(diamonds['cut'], categories=cut_order, ordered=True)
diamonds.stats.ordered_model(
    formula='cut ~ carat + price',
    distr='logit'
)
````

## Mixed Linear Model

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.mixedlm` - Linear regression with random effects.
```

````python
from frameon import load_dataset, FrameOn as fo

penguins = fo(load_dataset('penguins'))
penguins.stats.mixedlm(
    "body_mass_g ~ bill_length_mm + C(sex)",
    groups="island"
)
````

## Feature Importance Analysis

```{eval-rst}
:func:`~frameon.dataframe.statistics.FrameOnStats.feature_importance_analysis` - Analyzes feature importance for data analysis.
```

````python
from frameon import load_dataset, FrameOn as fo

iris = fo(load_dataset('iris'))
iris.stats.feature_importance_analysis(
    target_column='species',
    feature_columns=iris.columns.difference(['species', 'species_id']).tolist()
)
````
