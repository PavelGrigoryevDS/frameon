## Categorical Conversion

```{eval-rst}
:func:`~frameon.series.preprocessing.SeriesOnPreproc.to_categorical` - Convert numerical series to categorical using specified method.
``

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
rules = {
    "Child": lambda x: x <= 12,
    "Young": lambda x: x <= 25,
    "Adult": lambda x: x < 60,
    "Elderly": lambda x: x >= 60,
    "Missing Age": "default"
}
titanic['age_cat'] = titanic['age'].preproc.to_categorical(
    method='rules',
    rules=rules
)
````

## Text Processing

```{eval-rst}
:func:`~frameon.series.preprocessing.SeriesOnPreproc.normalize_string_series` - Normalize Series of strings with cleaning and standardization option.
```

````python
from frameon import FrameOn as fo
import pandas as pd

data = {
    "text": [
        "  Hello, how are you?  ",
        "RANDOM TEXT IN UPPERCASE",
        "café and naïve - special characters!",
        "one two.three,four",
        "example with 'quotes' and — dashes",
    ]
}
df = fo(pd.DataFrame(data))
df['cleaned_text'] = df['text'].preproc.normalize_string_series(
    case_format='title'
)
df
````

## Numeric Transformations

```{eval-rst}
:func:`~frameon.series.preprocessing.SeriesOnPreproc.transform_numeric` - Applies numeric transformations
```

````python
from frameon import load_dataset, FrameOn as fo

diamonds = fo(load_dataset('diamonds'))

diamonds['price_log'] = diamonds['price'].preproc.transform_numeric(
    method='boxcox',
    show_dist=True
)
````

## Missing Value Handling

```{eval-rst}
:func:`~frameon.series.preprocessing.SeriesOnPreproc.fill_missing_by_category` - Fills missing values by category
```

Fill missing values using category-based strategies.

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
print(f"Missings in age before handling: {titanic['age'].isna().sum()}")
titanic['age'] = titanic['age'].preproc.fill_missing_by_category(
    category_columns=['sex', 'class'],
    strategy='simple'
)
print(f"Missings in age after handling: {titanic['age'].isna().sum()}") 
````

```{eval-rst}
:func:`~frameon.series.preprocessing.SeriesOnPreproc.impute_missing` - Missing value imputation on specified numerical columns
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
print(f"Missings in age before handling: {titanic['age'].isna().sum()}")
titanic['age'] = titanic['age'].preproc.impute_missing(
    method='iterative',
)
print(f"Missings in age after handling: {titanic['age'].isna().sum()}") 
````

## Category Analysis

```{eval-rst}
:func:`~frameon.series.preprocessing.SeriesOnPreproc.calc_target_category_share` - Calculate the proportional share of a target category within grouped data, with support for time-based resampling.
```

````python
from frameon import load_dataset, FrameOn as fo

superstore = fo(load_dataset('superstore'))
los_angeles_share = superstore['City'].preproc.calc_target_category_share(
    target_category='Los Angeles',
    group_columns=['Order Date', 'Category'],
    resample_freq='ME',
)
los_angeles_share.head()
````
