## Info Method

```{eval-rst}
:func:`~frameon.dataframe.explore.FrameOnExplore.info` - Shows basic dataframe statistics
```

````python
from frameon import load_dataset, FrameOn as fo

iris = fo(load_dataset('iris'))
iris.explore.info()
````

## Anomalies Report

```{eval-rst}
:func:`~frameon.dataframe.explore.FrameOnExplore.anomalies_report` - Detects and reports anomalies
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.explore.anomalies_report(anomaly_type='missing')
````

## Detect Anomalies

```{eval-rst}
:func:`~frameon.dataframe.explore.FrameOnExplore.detect_anomalies` - Finds rows with anomalies
```

````python
from frameon import load_dataset, FrameOn as fo

tips = fo(load_dataset('tips'))
tips = fo(tips)
tips.explore.detect_anomalies(anomaly_type='outlier', method='quantile')
````

## Anomalies Correlation Matrix

```{eval-rst}
:func:`~frameon.dataframe.explore.FrameOnExplore.anomalies_corr_matrix` - Shows anomaly correlations
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
fig = titanic.explore.anomalies_corr_matrix(anomaly_type='missing')
fig.show()
````

## Anomalies Combinations


````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.explore.anomalies_combinations(anomaly_type='missing')
````

## Anomalies by Categories


````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.explore.anomalies_by_categories(anomaly_type='missing')
````

## Detect Simultaneous Anomalies

```{eval-rst}
:func:`~frameon.dataframe.explore.FrameOnExplore.detect_simultaneous_anomalies` - Finds rows with multiple anomalies
```

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
titanic.explore.detect_simultaneous_anomalies(
    anomaly_type='missing',
    columns=['age', 'deck']
)
````

## Anomalies Over Time

```{eval-rst}
:func:`~frameon.dataframe.explore.FrameOnExplore.anomalies_over_time` - Plot anomalies across ALL columns over time using resampling
```

````python
from frameon import load_dataset, FrameOn as fo

taxis = fo(load_dataset('taxis'))
fig = taxis.explore.anomalies_over_time(
    anomaly_type='missing',
    time_column='pickup'
)
fig.show()
````
