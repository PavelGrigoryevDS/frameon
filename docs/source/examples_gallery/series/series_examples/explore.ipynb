## Detect Anomalies

```{eval-rst}
:func:`~frameon.series.explore.SeriesOnExplore.detect_anomalies` - Detects anomalies in the series using the specified method.
```

Return boolean mask where True indicates anomalies in the series.

````python
from frameon import load_dataset, FrameOn as fo

titanic = fo(load_dataset('titanic'))
mask = titanic['age'].explore.detect_anomalies(
    anomaly_type='missing'
)
mask.head()
````

## Detect Outliers

```{eval-rst}
:func:`~frameon.series.explore.SeriesOnExplore.detect_outliers` - Detect outliers in series using statistical and machine learning methods.
```

````python
from frameon import load_dataset, FrameOn as fo

tips = fo(load_dataset('tips'))
tips['total_bill'].explore.detect_outliers(
    method='quantile',
    threshold=0.05
)
````

## Anomalies by Categories

```{eval-rst}
:func:`~frameon.series.explore.SeriesOnExplore.anomalies_by_categories` - Analyze anomaly distribution across all categorical columns in parent DataFrame.
```

````python
from frameon import load_dataset, FrameOn as fo

tips = fo(load_dataset('tips'))
tips['total_bill'].explore.anomalies_by_categories(
    anomaly_type='outlier',
    method='quantile',
    threshold=0.05
)
````

## Anomalies Over Time

```{eval-rst}
:func:`~frameon.series.explore.SeriesOnExplore.anomalies_over_time` - Plot anomalies over time using resampling.
```

````python
from frameon import load_dataset, FrameOn as fo

taxis = fo(load_dataset('taxis'))
fig = taxis['payment'].explore.anomalies_over_time(
    anomaly_type='missing',
    time_column='pickup',
    freq='1D'
)
fig.show()
````

## Detect Window Outliers

```{eval-rst}
:func:`~frameon.series.explore.SeriesOnExplore.detect_window_outliers` - Detect and analyze outliers in rolling windows of time series data.
```

````python
from frameon import load_dataset, FrameOn as fo

superstore = fo(load_dataset('superstore'))
superstore['Sales'].explore.detect_window_outliers(
    time_column='Order Date'
    , window=10
    , resample_freq='W'
    , agg_func='mean'
    , method='confidence'
    , threshold=0.05
)
````

## Plot Rolling Anomaly Rate

```{eval-rst}
:func:`~frameon.series.explore.SeriesOnExplore.plot_rolling_anomaly_rate` - Calculate and visualize the rolling rate of specified anomalies in a time series.
```

````python
from frameon import load_dataset, FrameOn as fo

taxis = fo(load_dataset('taxis'))
taxis['payment'].explore.plot_rolling_anomaly_rate(
    anomaly_type='missing',
    time_column='pickup',
    window=30, 
)
````
